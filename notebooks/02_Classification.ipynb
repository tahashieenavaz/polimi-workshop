```python
import numpy
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_openml, load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import ConfusionMatrixDisplay, classification_report
```

```python
__SEED = 1223334444
```

```python
X, y = fetch_openml("madelon", version=1, as_frame=True, return_X_y=True)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=__SEED
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(f"Dataset shape: {X.shape[0]} samples, {X.shape[1]} features")
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train_scaled, y_train)
score = model.score(X_test_scaled, y_test)
print(f"Accuracy Score: {score:.4f} ({score*100:.1f}%)")
```

```python
plt.figure(figsize=(8, 6))
ConfusionMatrixDisplay.from_estimator(
    model, X_test_scaled, y_test, cmap="Blues", colorbar=False
)
plt.title("Confusion Matrix: Where did the model get confused?")
plt.show()

print("\n--- Detailed Performance Report ---")
predictions = model.predict(X_test_scaled)
print(classification_report(y_test, predictions))
```

# Breast Cancer Dataset

```python
data = load_breast_cancer()

X = data.data
y = data.target
features = data.feature_names

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=__SEED
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(f"Dataset ready: {X.shape[0]} patients, {X.shape[1]} tumor measurements")
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train_scaled, y_train)

score = model.score(X_test_scaled, y_test)
print(f"\nAccuracy Score: {score:.4f} ({score*100:.1f}%)")
```

```python
plt.figure(figsize=(6, 5))
ConfusionMatrixDisplay.from_estimator(
    model,
    X_test_scaled,
    y_test,
    display_labels=data.target_names,
    cmap="Blues",
    colorbar=False,
)
plt.title("Confusion Matrix")
plt.show()
```

```python
coefficients = model.coef_[0]
top_indices = numpy.argsort(numpy.abs(coefficients))[-15:]
top_coeffs = coefficients[top_indices]
top_features = [features[i] for i in top_indices]

plt.figure(figsize=(10, 6))
colors = ["darkred" if c < 0 else "teal" for c in top_coeffs]
plt.barh(range(len(top_indices)), top_coeffs, color=colors)
plt.yticks(range(len(top_indices)), top_features)
plt.axvline(0, color="black", linewidth=1.5)  # Draw a line at zero
plt.xlabel("Impact on Prediction (Negative = Malignant, Positive = Benign)")
plt.title("Top 15 Features: How does the model diagnose?")
plt.grid(True, axis="x", linestyle="--")
plt.tight_layout()
plt.show()
```
