```python
import numpy
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
```

```python
__SEED = 323412
```

```python
def f(x):
    if 20 < x < 50:
        return x
    return x**2
```

```python
numpy.random.seed(__SEED)

X = numpy.random.uniform(0, 100, 150).reshape(-1, 1)
y = numpy.vectorize(f)(X) + numpy.random.normal(0, 5, size=(150, 1))
y = y.ravel()

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=__SEED
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
model = LinearRegression()
model.fit(X_train_scaled, y_train)

ensemble = RandomForestRegressor(n_estimators=100, random_state=__SEED)
ensemble.fit(X_train_scaled, y_train)

X_smooth = numpy.linspace(0, 100, 500).reshape(-1, 1)
X_smooth_scaled = scaler.transform(X_smooth)

model_line = model.predict(X_smooth_scaled)
ensemble_line = ensemble.predict(X_smooth_scaled)
```

```python
plt.figure(figsize=(14, 6))
plt.subplot(1, 2, 1)
plt.scatter(X_train, y_train, color="gray", alpha=0.5, label="Training Data")
plt.scatter(X_test, y_test, color="black", marker="x", label="Testing Data")
plt.plot(X_smooth, model_line, color="red", linewidth=3, label="Linear Prediction")
plt.title("Linear Regression (Underfitting)")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)

plt.subplot(1, 2, 2)
plt.scatter(X_train, y_train, color="gray", alpha=0.5, label="Training Data")
plt.scatter(X_test, y_test, color="black", marker="x", label="Testing Data")
plt.plot(
    X_smooth,
    ensemble_line,
    color="green",
    linewidth=3,
    label="Random Forest Prediction",
)
plt.title("Random Forest (Capturing Non-Linearity)")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)

plt.tight_layout()
plt.show()
```
