```python
import numpy
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.datasets import make_regression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures
```

```python
__SEED = 234124124
```

```python
X, y, true_coef = make_regression(
    n_samples=100,
    n_features=15,
    n_informative=3,
    noise=25.0,
    coef=True,
    random_state=__SEED
)

lr = LinearRegression()
ridge = Ridge(alpha=100.0)
lasso = Lasso(alpha=10.0)

lr.fit(X, y)
ridge.fit(X, y)
lasso.fit(X, y)
```

```python
plt.style.use('seaborn-v0_8-whitegrid')
fig = plt.figure(figsize=(12, 6))

features = numpy.arange(X.shape[1])
width = 0.25

plt.bar(features - width, lr.coef_, width, label='Linear (No Penalty)', color='#34495e')
plt.bar(features, ridge.coef_, width, label='Ridge ($L_2$ Penalty)', color='#3498db')
plt.bar(features + width, lasso.coef_, width, label='Lasso ($L_1$ Penalty)', color='#e74c3c')

plt.axhline(0, color='black', linestyle='--', linewidth=1)
plt.xlabel('Feature Index')
plt.ylabel('Coefficient Value / Weight')
plt.title('How Different Models Assign Importance to Features\n(Notice how Lasso completely drops the noise features)', fontsize=14)
plt.xticks(features)
plt.legend(fontsize=11)
plt.show()
```

```python
alphas = numpy.logspace(-2, 4, 100)
ridge_coefs = []
lasso_coefs = []

for a in alphas:
    ridge_path = Ridge(alpha=a).fit(X, y)
    ridge_coefs.append(ridge_path.coef_)

    lasso_path = Lasso(alpha=a).fit(X, y)
    lasso_coefs.append(lasso_path.coef_)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

ax1.plot(alphas, ridge_coefs)
ax1.set_xscale('log')
ax1.set_xlabel('Alpha Penalty (Log Scale)', fontsize=12)
ax1.set_ylabel('Coefficient Value', fontsize=12)
ax1.set_title('Ridge ($L_2$) Coefficient Paths\nWeights shrink, but rarely hit exactly zero', fontsize=13)

ax2.plot(alphas, lasso_coefs)
ax2.set_xscale('log')
ax2.set_xlabel('Alpha Penalty (Log Scale)', fontsize=12)
ax2.set_title('Lasso ($L_1$) Coefficient Paths\nWeights hit exactly zero (Feature Selection!)', fontsize=13)

plt.tight_layout()
plt.show()
```

```python
numpy.random.seed(__SEED)
X = numpy.sort(numpy.random.rand(30) * 2 * numpy.pi)
y = numpy.sin(X) + numpy.random.randn(30) * 0.3

X = X[:, numpy.newaxis]
X_plot = numpy.linspace(0, 2 * numpy.pi, 100)[:, numpy.newaxis]
```

```python
degree = 12

lr = make_pipeline(PolynomialFeatures(degree), LinearRegression())
ridge = make_pipeline(PolynomialFeatures(degree), Ridge(alpha=1e-2))
lasso = make_pipeline(PolynomialFeatures(degree), Lasso(alpha=1e-2, max_iter=10000))

lr.fit(X, y)
ridge.fit(X, y)
lasso.fit(X, y)
```

```python
plt.style.use('seaborn-v0_8-whitegrid')
plt.figure(figsize=(12, 7))

plt.scatter(X, y, color='black', s=40, zorder=5, label='Training Data (Noisy)')
plt.plot(X_plot, numpy.sin(X_plot), color='gray', linestyle='--', linewidth=2, label='True Underlying Function')
plt.plot(X_plot, lr.predict(X_plot), color='#2c3e50', linewidth=2.5,
         label='Linear Regression (Wildly Overfitted)')
plt.plot(X_plot, ridge.predict(X_plot), color='#3498db', linewidth=2.5,
         label='Ridge (Smoothed by L2)')
plt.plot(X_plot, lasso.predict(X_plot), color='#e74c3c', linewidth=2.5,
         label='Lasso (Smoothed by L1)')
plt.ylim(-2.5, 2.5)

plt.title('How Regularization Saves Us From Overfitting', fontsize=16, fontweight='bold')
plt.xlabel('Input Feature (X)', fontsize=12)
plt.ylabel('Target Variable (y)', fontsize=12)
plt.legend(fontsize=11, loc='upper right')

plt.tight_layout()
plt.show()
```
