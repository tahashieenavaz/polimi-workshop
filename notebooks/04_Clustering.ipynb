```python
import matplotlib.pyplot as plt
import random
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
```

```python
__SEED = 20034
```

```python
REAL_CENTERS = random.randint(7, 22)

X, y_true = make_blobs(
    n_samples=500, centers=REAL_CENTERS, cluster_std=1.2, random_state=__SEED
)

plt.figure(figsize=(8, 5))
plt.scatter(X[:, 0], X[:, 1], color="gray", alpha=0.5)
plt.title("Step 1: Raw, Unlabeled Data (What the computer sees)")
plt.show()
```

```python
k = 4
model = KMeans(n_clusters=k, random_state=__SEED, n_init="auto")
labels = model.fit_predict(X)
centers = model.cluster_centers_
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap="viridis", alpha=0.6)
plt.scatter(
    centers[:, 0],
    centers[:, 1],
    c="red",
    marker="X",
    s=200,
    linewidths=3,
    label="Centroids",
)
plt.title(f"Step 2: K-Means Clustering (Guessed K={k})")
plt.legend()
plt.show()
```

```python
inertias = []
k_range = range(1, 30)

for i in k_range:
    km = KMeans(n_clusters=i, random_state=__SEED, n_init="auto")
    km.fit(X)
    inertias.append(km.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(k_range, inertias, marker="o", color="teal", linewidth=2)
plt.title("Step 3: The Elbow Method (Finding the true K)")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia (How spread out the clusters are)")
plt.legend()
plt.grid(True, linestyle="--")
plt.show()
```
